```python
# 1. SETUP
!pip install kornia -q
import os, time, torch, torch.nn as nn, torch.nn.functional as F, torchvision, numpy as np
import torchvision.transforms as T, kornia.color as KC, matplotlib.pyplot as plt
from PIL import Image
from torch.utils.data import Dataset, DataLoader, ConcatDataset, Subset
from skimage.metrics import peak_signal_noise_ratio, structural_similarity

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.manual_seed(42)

IMG_SIZE, BATCH, EPOCHS, LR = 128, 32, 30, 2e-4
W_L1, W_PERC, W_GAN = 50.0, 10.0, 1.0      # loss weights
MAX_IMAGES = None                           # e.g. 3000 for a quick test
TEST_IMAGE = "/content/images.jpg"        # your own image

# RGB <-> LAB helpers (network works on L and ab scaled to [-1, 1])
def to_lab(rgb):
    lab = KC.rgb_to_lab(rgb)
    return lab[:, :1] / 50 - 1, lab[:, 1:] / 110          # L, ab

def to_rgb(L, ab):
    return KC.lab_to_rgb(torch.cat([(L + 1) * 50, ab * 110], 1)).clamp(0, 1)
print("device:", device)

# 2. DATA  (STL10: 13,000 natural images, split 90% train / 10% validation)
base = ConcatDataset([torchvision.datasets.STL10("./data", split=s, download=True) for s in ("train", "test")])
idx = torch.randperm(len(base)).tolist()[:MAX_IMAGES]
n_val = int(0.1 * len(idx))

class RGBData(Dataset):
    def __init__(self, ids, train):
        self.ds = Subset(base, ids)
        self.tf = T.Compose([T.Resize(IMG_SIZE), T.CenterCrop(IMG_SIZE)] + ([T.RandomHorizontalFlip()] if train else []) + [T.ToTensor()])
    def __len__(self): return len(self.ds)
    def __getitem__(self, i): return self.tf(self.ds[i][0].convert("RGB"))

train_loader = DataLoader(RGBData(idx[n_val:], True), BATCH, shuffle=True, num_workers=2, drop_last=True)
val_ds = RGBData(idx[:n_val], False)
val_loader = DataLoader(val_ds, BATCH, num_workers=2)
print("train:", len(idx) - n_val, "| validation:", n_val)


# 3. MODELS
def double_conv(i, o):
    return nn.Sequential(nn.Conv2d(i, o, 3, padding=1, bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                         nn.Conv2d(o, o, 3, padding=1, bias=False), nn.BatchNorm2d(o), nn.ReLU(True))

class UNet(nn.Module):                      # L (1 channel) -> ab (2 channels)
    def __init__(self, f=(32, 64, 128, 256)):
        super().__init__()
        self.pool = nn.MaxPool2d(2)
        self.down = nn.ModuleList(); c = 1
        for x in f: self.down.append(double_conv(c, x)); c = x
        self.mid = double_conv(f[-1], f[-1] * 2)
        self.up  = nn.ModuleList([nn.ConvTranspose2d(x * 2, x, 2, 2) for x in reversed(f)])
        self.dec = nn.ModuleList([double_conv(x * 2, x) for x in reversed(f)])
        self.out = nn.Conv2d(f[0], 2, 1)
    def forward(self, x):
        skips = []
        for d in self.down: x = d(x); skips.append(x); x = self.pool(x)
        x = self.mid(x)
        for up, dec, s in zip(self.up, self.dec, reversed(skips)):
            x = dec(torch.cat([s, up(x)], 1))          # skip connection
        return torch.tanh(self.out(x))

def PatchGAN():                             # judges (L, ab) patches as real / fake
    def b(i, o, s): return [nn.Conv2d(i, o, 4, s, 1), nn.BatchNorm2d(o), nn.LeakyReLU(0.2, True)]
    return nn.Sequential(nn.Conv2d(3, 64, 4, 2, 1), nn.LeakyReLU(0.2, True),
                         *b(64, 128, 2), *b(128, 256, 2), *b(256, 512, 1), nn.Conv2d(512, 1, 4, 1, 1))

class Perceptual(nn.Module):                # VGG16 feature distance
    def __init__(self):
        super().__init__()
        vgg = torchvision.models.vgg16(weights=torchvision.models.VGG16_Weights.IMAGENET1K_V1)
        self.f = vgg.features[:16].eval()
        for p in self.f.parameters(): p.requires_grad = False
        self.register_buffer("m", torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1))
        self.register_buffer("s", torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1))
    def forward(self, a, b): return F.l1_loss(self.f((a - self.m) / self.s), self.f((b - self.m) / self.s))

G, D, perc = UNet().to(device), PatchGAN().to(device), Perceptual().to(device)
bce = nn.BCEWithLogitsLoss()
print("U-Net parameters: %.1fM" % (sum(p.numel() for p in G.parameters()) / 1e6))

 # 4. TRAINING   Loss_G = 50*L1 + 10*Perceptual + 1*Adversarial
opt_G = torch.optim.Adam(G.parameters(), LR, betas=(0.5, 0.999))
opt_D = torch.optim.Adam(D.parameters(), LR, betas=(0.5, 0.999))
hist = {"G": [], "D": [], "val_psnr": []}

def psnr(a, b): return (10 * torch.log10(1 / ((a - b) ** 2).mean((1, 2, 3)).clamp(min=1e-10))).mean().item()

for epoch in range(1, EPOCHS + 1):
    G.train(); D.train(); t0 = time.time(); g_sum = d_sum = 0
    for rgb in train_loader:
        rgb = rgb.to(device); L, ab = to_lab(rgb)
        fake = G(L)

        # discriminator: real (L, ab) -> 1, fake (L, fake ab) -> 0
        p_real, p_fake = D(torch.cat([L, ab], 1)), D(torch.cat([L, fake.detach()], 1))
        loss_D = 0.5 * (bce(p_real, torch.ones_like(p_real)) + bce(p_fake, torch.zeros_like(p_fake)))
        opt_D.zero_grad(); loss_D.backward(); opt_D.step()

        # generator: L1 + perceptual + fool the discriminator
        p = D(torch.cat([L, fake], 1))
        loss_G = W_L1 * F.l1_loss(fake, ab) + W_PERC * perc(to_rgb(L, fake), rgb) + W_GAN * bce(p, torch.ones_like(p))
        opt_G.zero_grad(); loss_G.backward(); opt_G.step()
        g_sum += loss_G.item(); d_sum += loss_D.item()

    G.eval(); v = 0
    with torch.no_grad():
        for rgb in val_loader:
            rgb = rgb.to(device); L, _ = to_lab(rgb); v += psnr(to_rgb(L, G(L)), rgb) * len(rgb)
    hist["G"].append(g_sum / len(train_loader)); hist["D"].append(d_sum / len(train_loader)); hist["val_psnr"].append(v / n_val)
    print(f"Epoch {epoch:02d}/{EPOCHS} | G loss {hist['G'][-1]:.3f} | D loss {hist['D'][-1]:.3f} | val PSNR {hist['val_psnr'][-1]:.2f} dB | {time.time()-t0:.0f}s")

torch.save(G.state_dict(), "unet_colorizer.pth")

# 5. RESULTS: curves, PSNR / SSIM vs grayscale baseline, side-by-side images
fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
ax[0].plot(hist["G"], label="Generator"); ax[0].plot(hist["D"], label="Discriminator"); ax[0].set_title("Training loss"); ax[0].legend()
ax[1].plot(hist["val_psnr"], color="green"); ax[1].set_title("Validation PSNR (dB)"); plt.show()

G.eval(); P, S, Pg, Sg = [], [], [], []
with torch.no_grad():
    for rgb in val_loader:
        rgb = rgb.to(device); L, _ = to_lab(rgb)
        pred = to_rgb(L, G(L)).cpu().permute(0, 2, 3, 1).numpy()
        gray = to_rgb(L, torch.zeros(len(L), 2, IMG_SIZE, IMG_SIZE, device=device)).cpu().permute(0, 2, 3, 1).numpy()
        for p, g, t in zip(pred, gray, rgb.cpu().permute(0, 2, 3, 1).numpy()):
            P.append(peak_signal_noise_ratio(t, p, data_range=1)); S.append(structural_similarity(t, p, channel_axis=2, data_range=1))
            Pg.append(peak_signal_noise_ratio(t, g, data_range=1)); Sg.append(structural_similarity(t, g, channel_axis=2, data_range=1))
        if len(P) >= 300: break
print(f"Grayscale baseline : PSNR {np.mean(Pg):.2f} dB | SSIM {np.mean(Sg):.3f}")
print(f"U-Net colorization : PSNR {np.mean(P):.2f} dB | SSIM {np.mean(S):.3f}")


# 6. COLORIZE YOUR OWN IMAGE
if not os.path.exists(TEST_IMAGE):
    from google.colab import files
    TEST_IMAGE = "/content/" + list(files.upload().keys())[0]

img = Image.open(TEST_IMAGE).convert("RGB").resize((IMG_SIZE, IMG_SIZE))
x = T.ToTensor()(img).unsqueeze(0).to(device)
with torch.no_grad():
    L, _ = to_lab(x)
    out = to_rgb(L, G(L))
    gray = to_rgb(L, torch.zeros(1, 2, IMG_SIZE, IMG_SIZE, device=device))

fig, ax = plt.subplots(1, 3, figsize=(14, 5))
for a, (n, t) in zip(ax, [("Original", x), ("Grayscale input", gray), ("Colorized output", out)]):
    a.imshow(t[0].permute(1, 2, 0).cpu()); a.set_title(n); a.axis("off")
plt.show()
```
